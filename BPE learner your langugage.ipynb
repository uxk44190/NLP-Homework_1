```python
# ==========================================
# Q2.3 — Train BPE on your own paragraph
# Requirements covered:
# 1) Use end-of-word '_' and learn at least 30 merges
# 2) Show top 5 merges + 5 longest tokens
# 3) Segment 5 words (include 1 rare + 1 derived/inflected)
# ==========================================

from collections import Counter

EOW = "_"  # end-of-word marker


# ---------- Basic text -> word list ----------
def words_from_paragraph(paragraph: str):
    # Simple split (you can keep it simple for assignment)
    return [w.strip().lower() for w in paragraph.split() if w.strip()]


# ---------- Tokenization ----------
def tok_word(w: str):
    return list(w) + [EOW]


def tok_corpus(word_list):
    return [tok_word(w) for w in word_list]


# ---------- Bigram counting ----------
def bigram_counts(tokenized_words):
    c = Counter()
    for w in tokenized_words:
        c.update(zip(w, w[1:]))
    return c


# ---------- Merge operations ----------
def merge_word(tokens, pair):
    a, b = pair
    out, i = [], 0
    while i < len(tokens):
        if i < len(tokens) - 1 and tokens[i] == a and tokens[i + 1] == b:
            out.append(a + b)  # merge
            i += 2
        else:
            out.append(tokens[i])
            i += 1
    return out


def apply_merge(tokenized_words, pair):
    return [merge_word(w, pair) for w in tokenized_words]


def vocab_size(tokenized_words):
    return len({t for w in tokenized_words for t in w})


def token_set(tokenized_words):
    return {t for w in tokenized_words for t in w}


# ---------- BPE training ----------
def learn_bpe(paragraph: str, max_merges: int = 30, verbose: bool = True):
    wlist = words_from_paragraph(paragraph)
    corpus = tok_corpus(wlist)

    merges = []  # list of (pair, count_at_step)

    for step in range(1, max_merges + 1):
        bg = bigram_counts(corpus)
        if not bg:
            if verbose:
                print(f"Step {step:02d}: No bigrams left to merge.")
            break

        pair, cnt = bg.most_common(1)[0]
        corpus = apply_merge(corpus, pair)
        merges.append((pair, cnt))

        if verbose:
            print(
                f"Step {step:02d}: top pair = {pair} (count={cnt}) "
                f"-> vocab size = {vocab_size(corpus)}"
            )

    return merges, corpus, wlist


# ---------- Segment word using learned merges ----------
def segment(word: str, merges):
    tokens = tok_word(word.lower())
    for pair, _cnt in merges:
        tokens = merge_word(tokens, pair)
    return tokens


# ---------- Helper: top merges + longest tokens ----------
def top_k_merges(merges, k=5):
    # "most frequent merges" in BPE training = the chosen top pair at each step
    # We show first k steps (each was max at that step)
    return merges[:k]


def longest_k_tokens(final_corpus, k=5):
    toks = sorted(token_set(final_corpus), key=lambda t: (len(t), t), reverse=True)
    return toks[:k]


# ==========================================
# MAIN (edit paragraph + the 5 words if needed)
# ==========================================
if __name__ == "__main__":

    paragraph = (
        "Natural language processing helps computers understand human language. "
        "Subword tokenization is useful for handling rare and unseen words. "
        "Models learn patterns from text to improve understanding. "
        "Effective tokenization reduces vocabulary size and improves performance. "
        "This approach allows systems to generalize better to new word forms."
    )

    # 1) Train BPE (>= 30 merges)
    merges, final_corpus, word_list = learn_bpe(paragraph, max_merges=30, verbose=True)

    # 2) Show top 5 merges and 5 longest tokens
    print("\n=== Top 5 merges (pair -> merged_token) ===")
    for i, (pair, cnt) in enumerate(top_k_merges(merges, 5), 1):
        print(f"{i}. {pair} -> {pair[0] + pair[1]}   (count at step: {cnt})")

    print("\n=== 5 longest tokens in final vocabulary ===")
    for t in longest_k_tokens(final_corpus, 5):
        print(f"- {t} (len={len(t)})")

    # 3) Segment 5 words from the paragraph (include rare + derived/inflected)
    # Pick words from paragraph; you can change these to match your paragraph.
    # Example: rare word = "vocabulary" (likely occurs once)
    # derived form = "tokenization" or "generalize"
    test_words = ["language", "processing", "tokenization", "generalize", "vocabulary"]

    print("\n=== Segmentation of 5 words ===")
    for w in test_words:
        print(f"{w:<15} -> {' '.join(segment(w, merges))}")

    # Optional: show which word is rare in your paragraph (occurs once)
    freqs = Counter(word_list)
    rare = [w for w, c in freqs.items() if c == 1]
    print("\nRare words in paragraph (occur once):")
    print(", ".join(sorted(rare)) if rare else "None found")
```

```python

from collections import Counter
import re

EOW = "_"

def clean_words(text: str):
    text = re.sub(r"[^a-zA-Z\s]", " ", text).lower()   # remove punctuation
    return [w for w in text.split() if w]

def tok(w): return list(w) + [EOW]

def count_bigrams(words):
    c = Counter()
    for w in words: c.update(zip(w, w[1:]))
    return c

def merge_seq(seq, pair):
    a, b = pair
    out, i = [], 0
    while i < len(seq):
        if i < len(seq)-1 and seq[i] == a and seq[i+1] == b:
            out.append(a+b); i += 2
        else:
            out.append(seq[i]); i += 1
    return out

def learn_bpe(paragraph, n_merges=30):
    corpus = [tok(w) for w in clean_words(paragraph)]
    merges = []
    for s in range(1, n_merges+1):
        bg = count_bigrams(corpus)
        if not bg:
            print(f"Step {s:02d}: No bigrams left to merge."); break
        pair, cnt = bg.most_common(1)[0]
        corpus = [merge_seq(w, pair) for w in corpus]
        merges.append((pair, cnt))
        vocab = {t for w in corpus for t in w}
        print(f"Step {s:02d}: top pair = {pair} (count={cnt}) -> vocab size = {len(vocab)}")
    return merges, corpus, Counter(clean_words(paragraph))

def segment(word, merges):
    seq = tok(re.sub(r"[^a-zA-Z]", "", word).lower())
    for (pair, _cnt) in merges:
        seq = merge_seq(seq, pair)
    return seq

if __name__ == "__main__":
    paragraph = (
        "Natural language processing helps computers understand human language. "
        "Subword tokenization is useful for handling rare and unseen words. "
        "Models learn patterns from text to improve understanding. "
        "Effective tokenization reduces vocabulary size and improves performance. "
        "This approach allows systems to generalize better to new word forms."
    )

    merges, final_corpus, wf = learn_bpe(paragraph, n_merges=30)

    print("\n=== Top 5 merges ===")
    for i, (pair, cnt) in enumerate(merges[:5], 1):
        print(f"{i}. {pair} -> {pair[0]+pair[1]} (count={cnt})")

    print("\n=== 5 longest tokens ===")
    toks = sorted({t for w in final_corpus for t in w}, key=lambda x: (len(x), x), reverse=True)[:5]
    for t in toks:
        print(f"- {t} (len={len(t)})")

    # Choose 5 words (one rare + one derived/inflected). Edit if needed.
    test_words = ["language", "processing", "tokenization", "generalize", "vocabulary"]

    print("\n=== Segmentation of 5 words ===")
    for w in test_words:
        print(f"{w:<15} -> {' '.join(segment(w, merges))}")

    rare = sorted([w for w, c in wf.items() if c == 1])
    print("\nRare words (occur once):")
    print(", ".join(rare) if rare else "None found")
```
